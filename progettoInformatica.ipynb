# Add to the dataframe the following columns:

pass_rate: the ratio between units_passed and units_enrolled;
gpa_change: the difference between gpa_semester_2 and gpa_semester_1;
gpa_trend: a string that is "improving" if gpa_change is larger than 0.5, "declining" if gpa_change is smaller than -0.5, and "stable" otherwise.

```python
# Add to the dataframe the following columns:

# pass_rate: the ratio between units_passed and units_enrolled;
# gpa_change: the difference between gpa_semester_2 and gpa_semester_1;
# gpa_trend: a string that is "improving" if gpa_change is larger than 0.5, "declining" if gpa_change is smaller than -0.5, and "stable" otherwise.

student_dropout['pass_rate'] = student_dropout['units_passed'] / student_dropout['units_enrolled']
student_dropout['gpa_change'] = student_dropout['gpa_semester_2'] - student_dropout['gpa_semester_1']
```

```python
def gpa_trend_funz(gpa_change):
    if gpa_change > 0.5:
        return 'improving'
    if gpa_change < -0.5:
        return 'declining'
    else:
        return 'stable'


# poi provo a creare una nuova colonna del dataframe student_dropout chiamata gpa_trend con come records i valori risultanti dalla funzione applicata su gpa_change
student_dropout['gpa_trend'] = student_dropout['gpa_change'].apply(gpa_trend_funz)
```

# Partition the students into the following groups, according to weekly_work_hours: 0, 1-10, 11-20, 21-30, more than 30 (all intervals include both extremes).

```python
# 6:
def get_work_group(hours):
    if hours == 0:
        return '0'
    elif 1 <= hours <= 10:
        return '1-10'
    elif 11 <= hours <= 20:
        return '11-20'
    elif 21 <= hours <= 30:
        return '21-30'
    else:
        return 'more than 30'


print(student_dropout['weekly_work_hours'].apply(get_work_group))
```

Build a risk score by summing:

2 points if the tuition is not up to date;
1 point if stress_level is at least 8;
1 point if sleep_hours is smaller than 5;
1 point if the student has a part-time job with more than 20 weekly work hours;
2 points if attendance_rate is smaller than 60;
1 point if missed_deadlines is larger than 3.


```python
#7. Build a risk score by summing:

# 2 points if the tuition is not up to date;
# 1 point if stress_level is at least 8;
# 1 point if sleep_hours is smaller than 5;
# 1 point if the student has a part-time job with more than 20 weekly work hours;
# 2 points if attendance_rate is smaller than 60;
# 1 point if missed_deadlines is larger than 3.
```

```python
def funz_risk_score(riga):
    punti = 0 
    
    # 2 punti se la rata non è ancora stata pagata
    if riga['tuition_up_to_date'] == 0:
        punti += 2
    
    # 1 punto se stress_level eè almeno 8
    if riga['stress_level'] >= 8:
        punti += 1
        
    # 1 punto se sleep_hours è più piccolo di 5
    if riga['sleep_hours']< 5:
        punti += 1
    
    # 1 punto se lo studente ha part_time_job e weekly_work_hours > 20
    if riga['part_time_job'] == 1 and riga['weekly_work_hours'] > 20: 
        punti += 1 
    
    # 2 points if attendance_rate is smaller than 60 
    
    if riga['attendance_rate']< 60:
        punti += 2
        
    # 1 punto se missed_deadlines is larger than 3
    if riga['missed_deadlines'] > 3: 
        punti += 1
        
    return punti 


# applico la funzione a tutto il dataframe e assegno una nuova colonna risk_score al dataframe 

student_dropout['risk_score'] = student_dropout.apply(funz_risk_score, axis = 1) # a tutte le riga del dataframe
```

#Clean the dataset, removing the rows that do not correspond to a single country (e.g. World, High-income countries). For each country keep only the row with the most recent year.



```python
pattern = r"^(?:[Ww]orld|[Ll]ow-income countries|[Hh]igh-income countries|[Ll]ower-middle-income countries|[Uu]pper-middle-income countries|Middle East, North Africa, Afghanistan and Pakistan \(WB\)|Europe and Central Asia \(WB\)|East Asia and Pacific \(WB\))$"

# Per visualizzare i record trovati:
trovati = gdp[gdp["entity"].str.contains(pattern, regex=True, na=False)]
# na = False significa che i valori nulli li devo considerare esclusi dalla colonna

# Per eliminarli dal dataset:
gdp_clean = gdp[~gdp["entity"].str.contains(pattern, regex=True, na=False)].copy()
# il dataframe pulito ci servirà nei passaggi seguenti

gdp_clean["entity"]
```
